## **15.3 - Convert XML Annotations to YOLO Format**

````python
import os
import xml.etree.ElementTree as ET

# Source annotation folder
annotation_dir = os.path.join(path, "annotations")

# Output label folder
label_dir = "/content/yolo_dataset/labels"

os.makedirs(label_dir, exist_ok=True)

# Class Mapping
classes = {"pothole": 0}

def convert_bbox(size, box):
    dw = 1.0 / size[0]
    dh = 1.0 / size[1]

    x = (box[0] + box[1]) / 2.0
    y = (box[2] + box[3]) / 2.0

    w = box[1] - box[0]
    h = box[3] - box[2]

    x *= dw
    y *= dh
    w *= dw
    h *= dh

    return (x, y, w, h)

for xml_file in os.listdir(annotation_dir):

    tree = ET.parse(os.path.join(annotation_dir, xml_file))
    root = tree.getroot()

    width = int(root.find("size/width").text)
    height = int(root.find("size/height").text)

    txt_name = xml_file.replace(".xml", ".txt")

    with open(os.path.join(label_dir, txt_name), "w") as f:

        for obj in root.findall("object"):

            cls = obj.find("name").text

            if cls not in classes:
                continue

            xmlbox = obj.find("bndbox")

            xmin = float(xmlbox.find("xmin").text)
            xmax = float(xmlbox.find("xmax").text)
            ymin = float(xmlbox.find("ymin").text)
            ymax = float(xmlbox.find("ymax").text)

            bbox = convert_bbox(
                (width, height),
                (xmin, xmax, ymin, ymax)
            )

            f.write(
                f"{classes[cls]} {' '.join(map(str, bbox))}\n"
            )

print("✅ XML annotations successfully converted to YOLO format.")
````

## **15.6 - Split Dataset(Train/Validation/Test)**

````python
import os
import random
import shutil
from sklearn.model_selection import train_test_split

# Source folders
image_dir = os.path.join(path, "images")
label_dir = "/content/yolo_dataset/labels"

# Destination folders
train_img = "/content/yolo_dataset/train/images"
train_lbl = "/content/yolo_dataset/train/labels"

valid_img = "/content/yolo_dataset/valid/images"
valid_lbl = "/content/yolo_dataset/valid/labels"

test_img = "/content/yolo_dataset/test/images"
test_lbl = "/content/yolo_dataset/test/labels"

# Image list
images = [f for f in os.listdir(image_dir) if f.endswith(".png")]

# Split 80% Train, 10% Validation, 10% Test
train_files, temp_files = train_test_split(
    images,
    test_size=0.20,
    random_state=42
)

valid_files, test_files = train_test_split(
    temp_files,
    test_size=0.50,
    random_state=42
)

def copy_files(file_list, img_dest, lbl_dest):
    for img in file_list:

        shutil.copy(
            os.path.join(image_dir, img),
            os.path.join(img_dest, img)
        )

        label = img.replace(".png", ".txt")

        shutil.copy(
            os.path.join(label_dir, label),
            os.path.join(lbl_dest, label)
        )

copy_files(train_files, train_img, train_lbl)
copy_files(valid_files, valid_img, valid_lbl)
copy_files(test_files, test_img, test_lbl)

print("✅ Dataset Split Completed Successfully.")
````

# **17. Training Strategy**

The YOLOv8 Nano model will be trained using the prepared pothole detection dataset. The dataset has already been converted into YOLO format and split into training, validation, and testing sets.
